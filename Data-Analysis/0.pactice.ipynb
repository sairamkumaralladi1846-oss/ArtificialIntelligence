```python
import numpy as np
import time
import sys
```

```python
n = 1_000_000

list_a = [i for i in range(n)]
start = time.time()

list_b = [2*i for i in list_a]
end = time.time()
print("Time taken:", end - start)
```

```python
arr_a = np.array(list_a)
start = time.time()
arr_b = 2 * arr_a
end = time.time()
print("Time taken with NumPy:", end - start)
```

```python
print(f'Size of a Python list: {sys.getsizeof(list_a)} bytes')
print(f'Size of a NumPy array: {arr_a.nbytes} bytes')
```
