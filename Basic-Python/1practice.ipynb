## Read the Data from the JSON file

```python
with open("../Data-Analysis/data/store_data.json", "r") as f:
    data = json.load(f)
```

```python
type(data[0])
```

```python
text_num_dict = {
    "one": 1,
    "two": 2,
    "three": 3, 
    "four": 4,  
    "five": 5
}
```

```python
def get_number_from_text(text):
    return text_num_dict.get(text.lower(), None)
```

```python
def is_number(n):
    try:
        float(n)
        return True
    except:
        return False
```

```python
cleaned_data = []

for item in data:
    dict_item = {}
    dict_item["name"] = item.get("name", None)
    
    age = item.get("age", None)
    if is_number(age):
        age = int(age)  
    dict_item["age"] = age   

    rating = item.get("rating", None)
    if is_number(rating):
        if float(rating) == int(float(rating)):
            rating = int(float(rating.strip()))
        else:
            rating = float(rating.strip())
    else:
        rating = get_number_from_text(rating) 
    dict_item["rating"] = rating 

    dict_item['feedback'] = item.get('feedback', None)
    cleaned_data.append(dict_item)
```
