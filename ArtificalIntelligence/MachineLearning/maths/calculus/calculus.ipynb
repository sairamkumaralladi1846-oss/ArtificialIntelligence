```python
# calculus is the study of continuous change, and it has two main branches: differential calculus and integral calculus. 
# Differential calculus focuses on the concept of the derivative, which represents the rate of change of a function. 
# Integral calculus, on the other hand, deals with the concept of the integral, which represents the accumulation of quantities and the area under curves.
# Functions are mathematical relationships between inputs and outputs, and they can be represented in various forms, such as equations, graphs, or tables.
# Domain and range are important concepts in calculus. The domain of a function is the set of all possible input values (x-values) for which the function is defined
# while the range is the set of all possible output values (y-values) that the function can produce.
# Represented if [] : meaning include, () : meaning exclude.
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Example platting th sqrt(x) function
# Domain [0, infinity) 
# Range [0, infinity)

x = np.linspace(0,100, 1000)
y = np.sqrt(x)

plt.plot(x, y, color='blue')
plt.xlabel('x')
plt.ylabel('sqrt(x)')
plt.title('Plot of sqrt(x) function')
plt.legend(['sqrt(x)'])
plt.show()
```

```python
# function sinx
# Domain (-infinity, infinity)
# Range [-1, 1]

x = np.linspace(-2*np.pi, 2*np.pi, 1000)
y = np.sin(x)

plt.axvline(0, color='black', lw=1.5, linestyle='--')
plt.axhline(0, color='black', lw=1.5, linestyle='--')
plt.plot(x, y, color='blue')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Plot of sin(x) function')
plt.legend(['sin(x)'])
plt.show()
```

```python
# Scalar multiplication of a function
# f(x) = 2sin(x), 1/2sin(x)
# Domain (-infinity, infinity)
# Range:[-1, 1], [-2, 2], [-0.5, 0.5], [-2, 2]

x = np.linspace(-2*np.pi, 2*np.pi, 1000)
y = np.sin(x)
y1 = 2 * np.sin(x)
y2 = 0.5 * np.sin(x)
y3 = -2 * np.sin(x)

plt.axvline(0, color='black', lw=1.5, linestyle='--')
plt.axhline(0, color='black', lw=1.5, linestyle='--')
plt.plot(x, y, color='blue', label='sin(x)')
plt.plot(x, y1, color='red', label='2sin(x)')
plt.plot(x, y2, color='green', label='0.5sin(x)')
plt.plot(x, y3, color='orange', label='-2sin(x)')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of 2sin(x) function')
plt.xticks(np.arange(-2*np.pi, 2*np.pi+0.5, 0.5*np.pi),
           ['-2π', '-3π/2', '-π', '-π/2', '0', 'π/2', 'π', '3π/2', '2π'])
plt.yticks(np.arange(-2, 2.5, 0.5))
plt.legend(loc='upper right')
plt.show()
plt.tight_layout()
```

```python
# Scalar Addition of a function
# f(x) = sin(x) + 2, sin(x) - 2
# Domain (-infinity, infinity)
# Range: [-1, 1], [1, 3], [-3, -1]

x = np.linspace(-2*np.pi, 2*np.pi, 1000)
y = np.sin(x)
y1 = np.sin(x) + 2
y2 = np.sin(x) - 2

plt.axvline(0, color='black', lw=1.5, linestyle='--')
plt.axhline(0, color='black', lw=1.5, linestyle='--')
plt.plot(x, y, color='blue', label='sin(x)')
plt.plot(x, y1, color='red', label='sin(x) + 2')
plt.plot(x, y2, color='green', label='sin(x) - 2')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of sin(x) + 2 and sin(x) - 2 functions')
plt.xticks(np.arange(-2*np.pi, 2*np.pi+0.5, 0.5*np.pi),
           ['-2π', '-3π/2', '-π', '-π/2', '0', 'π/2', 'π', '3π/2', '2π'])
plt.yticks(np.arange(-3, 4, 1))
plt.legend(loc='upper right')
plt.show()
```

```python
# Scalar multiplication of input variable
# f(x) = sin(2x), sin(0.5x)
# Domain (-infinity, infinity)
# Range: [-1, 1]

x = np.linspace(-2*np.pi, 2*np.pi, 1000)
y = np.sin(x)
y1 = np.sin(2*x)
y2 = np.sin(0.5*x)

plt.axvline(0, color='black', lw=1.5, linestyle='--')
plt.axhline(0, color='black', lw=1.5, linestyle='--')
plt.plot(x, y, color='blue', label='sin(x)')
plt.plot(x, y1, color='red', label='sin(2x)')
plt.plot(x, y2, color='green', label='sin(0.5x)')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of sin(2x) and sin(0.5x) functions')
plt.xticks(np.arange(-2*np.pi, 2*np.pi+0.5, 0.5*np.pi),
           ['-2π', '-3π/2', '-π', '-π/2', '0', 'π/2', 'π', '3π/2', '2π'])
plt.yticks(np.arange(-1, 1.5, 0.5))
plt.legend(loc='upper right')   
```

```python
#Scalar addition of input variable
# f(x) = sin(x + 2), sin(x - 2)
# Domain (-infinity, infinity)
# Range: [-1, 1]

x = np.linspace(-2*np.pi, 2*np.pi, 1000)
y = np.sin(x)
y1 = np.sin(x + 2)
y2 = np.sin(x - 2)

plt.axvline(0, color='black', lw=1.5, linestyle='--')
plt.axhline(0, color='black', lw=1.5, linestyle='--')
plt.plot(x, y, color='blue', label='sin(x)')
plt.plot(x, y1, color='red', label='sin(x + 2)')
plt.plot(x, y2, color='green', label='sin(x - 2)')  


plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of sin(x + 2) and sin(x - 2) functions  ')
plt.xticks(np.arange(-2*np.pi, 2*np.pi+0.5, 0.5*np.pi),
           ['-2π', '-3π/2', '-π', '-π/2', '0', 'π/2', 'π', '3π/2', '2π'])
plt.yticks(np.arange(-1, 1.5, 0.5))
plt.legend(loc='upper right')
```
