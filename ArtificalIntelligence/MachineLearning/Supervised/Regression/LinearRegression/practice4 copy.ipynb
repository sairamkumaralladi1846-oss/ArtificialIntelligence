```python
x = [1,2,3,4,5]
y = [2,4,6,8,11]
print("x:", x)
print("y:", y)
```

```python
theta = [-5,-4,-3,-2,-1,0,1,2,2.025,2.05,2.1,2.2,2.3,2.4,2.5,2.6,2.7,2.8,2.9,3,4,5]
# 1/2*m sum((y_pred - y)**2)
j_ = []
```

```python
for i in theta:
    y_pred = [i*j for j in x]
    loss = 1/(2*len(x)) * sum([(y_pred[j] - y[j])**2 for j in range(len(y))])
    j_.append(loss)
    print(f"Theta: {i}, Loss: {loss}")


```

```python
import matplotlib.pyplot as plt

plt.plot(theta, j_, marker = '>')
plt.xlabel("Theta")
plt.ylabel("Loss")
plt.title("Loss vs Theta")
plt.show()
```

```python
lam = 0.2
j_ridge = []
for i in theta:
    y_pred = [i*j for j in x]
    loss = 1/(2*len(x)) * sum([(y_pred[j] - y[j])**2 for j in range(len(y))]) + lam * i**2   # penalty uses current theta
    j_ridge.append(loss)
    print(f"Theta: {i}, Loss: {loss}")

print("Best theta without penalty:", theta[j_.index(min(j_))])
print("Best theta with ridge     :", theta[j_ridge.index(min(j_ridge))])
```
